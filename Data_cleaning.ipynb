### Filtering Out Feature Phones and Non-Smartphone Devices
We will identify and remove basic feature phones based on RAM or internal storage (ROM) reported in `MB` instead of `GB` (or missing OS, CPU, and RAM simultaneously), while also filtering out tablets, smartwatches, and rows with currency unit errors or extreme luxury outliers to avoid distorting subsequent analyses.

```python
import re
import pandas as pd

def is_non_target_smartphone(row):
    ram_str = str(row['RAM']).upper() if pd.notna(row['RAM']) else ''
    rom_str = str(row['internal_memory']).upper() if pd.notna(row['internal_memory']) else ''

    # 1. Filter out Feature Phones (basic phones)
    if 'MB' in ram_str or 'MB' in rom_str or '512 ROM' in rom_str:
        return True
    if pd.isna(row['OS']) and pd.isna(row['CPU']) and pd.isna(row['RAM']):
        return True

    # 2. Filter out Smartwatches, Tablets, and non-cellular devices
    os_str = str(row['OS']).upper() if pd.notna(row['OS']) else ''
    model_str = str(row['model']).upper()

    if 'WATCH' in os_str or 'WATCH' in model_str:
        return True
    if str(row['network_technology']).strip().upper() == 'NO CELLULAR CONNECTIVITY':
        return True
    if any(k in model_str for k in ['IPAD', 'TAB ', 'TAB-', 'TABLET', 'TOUGHPAD', 'FUNBOOK', 'STREAK', 'SPRINGBOARD']):
        return True

    # 3. Screen size filter: Scan both columns because their contents are swapped in the raw dataset
    res_str = str(row['display_resolution']) if pd.notna(row['display_resolution']) else ''
    size_str = str(row['display_size']) if pd.notna(row['display_size']) else ''
    combined_display_info = f"{res_str} {size_str}"

    m = re.search(r'(\d+(?:\.\d+)?)\s*inches', combined_display_info, re.IGNORECASE)
    if m:
        size_inch = float(m.group(1))
        # Keep the upper bound at 8.0 inches to retain foldable smartphones (e.g., Z Fold at 7.6 inches)
        if size_inch > 8.0 or size_inch < 3.0:
            return True

    # 4. Filter out extreme price outliers and luxury handcrafted phones
    if row['approx_price_EUR'] > 2500 or str(row['brand']).strip().upper() == 'VERTU':
        return True

    return False

# Apply the filter
initial_count = len(df)
df = df[~df.apply(is_non_target_smartphone, axis=1)].copy()

print(f"Initial number of devices: {initial_count}")
print(f"Number of devices after filtering: {len(df)}")
```

**The next section will extract the key technical specifications of a smartphone**

### 1. Processing `RAM` and `internal_memory` Columns to Extract RAM and ROM

```python
import re
import pandas as pd
import numpy as np

def extract_ram_rom(row):
    ram_val = row['RAM']
    rom_val = row['internal_memory']

    # Extract RAM capacity (e.g., "2 GB RAM", "3/4 GB RAM")
    ram = None
    if pd.notna(ram_val):
        s_ram = str(ram_val).upper()
        m_ram = re.findall(r'(?<![A-Z0-9])(\d+(?:\.\d+)?)\s*GB\s*RAM', s_ram)
        if not m_ram:
            m_ram = re.findall(r'(?<![A-Z0-9])(\d+(?:\.\d+)?)\s*GB', s_ram)
        vals = [float(x) for x in m_ram if float(x) <= 16]
        if vals:
            ram = f"{vals[0]} GB"

    # Extract internal storage ROM capacity (e.g., "16 GB", "16/32 GB")
    rom = None
    if pd.notna(rom_val):
        s_rom = str(rom_val).upper().strip()
        m_rom = re.findall(r'(?<![A-Z0-9])(\d+(?:\.\d+)?)\s*GB', s_rom)
        if m_rom:
            rom = f"{float(m_rom[0])} GB"
        elif re.match(r'^\d+(?:\.\d+)?$', s_rom):
            rom = f"{float(s_rom)} GB"

    return ram, rom

# Apply the function to standardize the RAM column and create the ROM column
ram_rom_extracted = df.apply(extract_ram_rom, axis=1)
df['RAM_Clean'] = [x[0] for x in ram_rom_extracted]
df['ROM'] = [x[1] for x in ram_rom_extracted]

display(df[['RAM', 'internal_memory', 'RAM_Clean', 'ROM']].head())
```

### 2. Processing Display Columns (`display_resolution`, `display_type`, `display_size`) to Extract Screen Size, Screen-to-Body Ratio, Display Panel, and Display Resolution
*(Note: In the raw dataset, the `display_resolution` and `display_size` column names are swapped — `display_resolution` contains the physical size in inches along with the screen-to-body ratio, while `display_size` contains the pixel resolution).*

```python
def extract_screen_info(row):
    res_col = str(row['display_resolution']) if pd.notna(row['display_resolution']) else ""
    size_col = str(row['display_size']) if pd.notna(row['display_size']) else ""
    type_col = str(row['display_type']) if pd.notna(row['display_type']) else ""

    # 1. Screen size: accompanied by the word "inches"
    size_match = re.search(r'(\d+(?:\.\d+)?)\s*inches', res_col, re.IGNORECASE)
    size = f"{size_match.group(1)} inch" if size_match else None

    # 2. Screen-to-body ratio: accompanied by the "%" symbol
    ratio_match = re.search(r'(\d+(?:\.\d+)?)\s*%', res_col)
    screen_ratio = f"{ratio_match.group(1)} %" if ratio_match else None

    # 3. Display Panel (Display technology): Super AMOLED, AMOLED, IPS LCD, Super LCD, TFT LCD...
    t_up = type_col.upper()
    if 'SUPER AMOLED' in t_up:
        panel = 'Super AMOLED'
    elif 'AMOLED' in t_up:
        panel = 'AMOLED'
    elif 'OLED' in t_up:
        panel = 'OLED'
    elif 'IPS' in t_up:
        panel = 'IPS LCD'
    elif 'SUPER LCD' in t_up or 'S-LCD' in t_up:
        panel = 'Super LCD'
    elif 'TFT' in t_up:
        panel = 'TFT LCD'
    elif 'PLS' in t_up:
        panel = 'PLS LCD'
    else:
        panel = None

    # 4. Display resolution: e.g., 720 x 1280 pixels
    res_match = re.search(r'(\d+\s*[xX]\s*\d+\s*(?:Pixels|pixels)?)', size_col)
    resolution = res_match.group(1).strip() if res_match else None

    return size, screen_ratio, panel, resolution

# Apply the function to extract display information
screen_extracted = df.apply(extract_screen_info, axis=1)
df['Screen Size'] = [x[0] for x in screen_extracted]
df['Screen-to-Body Ratio'] = [x[1] for x in screen_extracted]
df['Display Panel'] = [x[2] for x in screen_extracted]
df['Display Resolution'] = [x[3] for x in screen_extracted]

display(df[['display_resolution', 'display_type', 'Screen Size', 'Screen-to-Body Ratio', 'Display Panel', 'Display Resolution']].head(10))
```

### 3. Processing `primary_camera` and `secondary_camera` Columns to Extract Rear and Front Camera Resolution (MP)

```python
def extract_camera_info(row):
    p_str = str(row['primary_camera']).upper() if pd.notna(row['primary_camera']) else ""
    s_str = str(row['secondary_camera']).upper() if pd.notna(row['secondary_camera']) else ""

    # Extract the highest Rear Camera resolution
    mp_rear = re.findall(r'(?<![A-Z\d])(\d+(?:\.\d+)?)\s*MP', p_str)
    vals_p = [float(val) for val in mp_rear if float(val) <= 100]
    if vals_p:
        rear_cam = f"{max(vals_p)} MP"
    elif 'VGA' in p_str:
        rear_cam = "0.3 MP"
    else:
        rear_cam = None

    # Extract Front Camera resolution (if 'No', assign 0 MP; if 'VGA', assign 0.3 MP)
    if s_str.strip() == 'NO':
        front_cam = "0 MP"
    else:
        mp_front = re.findall(r'(?<![A-Z\d])(\d+(?:\.\d+)?)\s*MP', s_str)
        vals_s = [float(val) for val in mp_front if float(val) <= 100]
        if vals_s:
            front_cam = f"{max(vals_s)} MP"
        elif 'VGA' in s_str:
            front_cam = "0.3 MP"
        else:
            front_cam = None

    return rear_cam, front_cam

# Apply the function to extract rear and front camera resolutions
camera_extracted = df.apply(extract_camera_info, axis=1)
df['Rear Camera'] = [x[0] for x in camera_extracted]
df['Front Camera'] = [x[1] for x in camera_extracted]

display(df[['primary_camera', 'secondary_camera', 'Rear Camera', 'Front Camera']].head(10))
```

### 4. Processing `battery` Column to Extract Battery Capacity (mAh) and Battery Design (Non-removable / Removable)

```python
def extract_battery_info(text):
    if pd.isna(text):
        return None, None
    text_str = str(text)

    # Find battery capacity (e.g., 3400 mAh)
    battery_match = re.search(r'(\d+)\s*mAh', text_str, re.IGNORECASE)
    battery_capacity = f"{battery_match.group(1)} mAh" if battery_match else None

    # Determine whether the battery is Non-removable or Removable
    if 'NON-REMOVABLE' in text_str.upper():
        battery_design = 'Non-removable'
    else:
        battery_design = 'Removable'

    return battery_capacity, battery_design

# Apply the function to extract battery capacity and battery design
battery_extracted = df['battery'].apply(extract_battery_info)
df['Battery Capacity'] = [x[0] for x in battery_extracted]
df['Battery Design'] = [x[1] for x in battery_extracted]

display(df[['battery', 'Battery Capacity', 'Battery Design']].head(10))
```

### 5. Processing Dimensions (`dimentions`), Weight (`weight_g`), and Checking Fingerprint Sensor (`sensors`)

```python
def extract_design_and_sensors(row):
    dim_text = str(row['dimentions']) if pd.notna(row['dimentions']) else ""
    sensors_text = str(row['sensors']) if pd.notna(row['sensors']) else ""

    # 1. Extract device thickness (mm) from the Length x Width x Thickness mm string
    thick_match = re.search(r'x\s*\d+(?:\.\d+)?\s*x\s*(\d+(?:\.\d+)?)\s*mm', dim_text, re.IGNORECASE)
    if not thick_match:
        thick_match = re.search(r'(\d+(?:\.\d+)?)\s*mm\s*thickness', dim_text, re.IGNORECASE)
    thickness = f"{thick_match.group(1)} mm" if thick_match else None

    # 2. Standardize device weight (g)
    w_val = pd.to_numeric(row['weight_g'], errors='coerce')
    weight = f"{w_val} g" if pd.notna(w_val) else None

    # 3. Check for Fingerprint sensor support
    if re.search(r'fingerprint', sensors_text, re.IGNORECASE):
        fingerprint = "Available"
    else:
        fingerprint = "No"

    return thickness, weight, fingerprint

# Apply the function to extract thickness, weight, and fingerprint sensor
design_extracted = df.apply(extract_design_and_sensors, axis=1)
df['Thickness'] = [x[0] for x in design_extracted]
df['Weight'] = [x[1] for x in design_extracted]
df['Fingerprint Sensor'] = [x[2] for x in design_extracted]

display(df[['dimentions', 'weight_g', 'sensors', 'Thickness', 'Weight', 'Fingerprint Sensor']].head(10))
```

### 6. Checking Cellular Network Connectivity Support (2G, 3G, or 4G LTE)

```python
def check_network_support(row):
    net_tech = str(row['network_technology']) if pd.notna(row['network_technology']) else ""
    bands_4g = str(row['4G_bands']) if pd.notna(row['4G_bands']) else ""
    combined_text = (net_tech + " " + bands_4g).upper()

    # Check for the highest cellular network generation supported by the device
    if "LTE" in combined_text or "4G" in combined_text:
        return "4G / LTE"
    elif any(k in combined_text for k in ["HSPA", "3G", "UMTS", "EVDO"]):
        return "3G"
    else:
        return "2G"

# Apply the function to determine supported network generation
df['Network Support'] = df.apply(check_network_support, axis=1)

display(df[['network_technology', '4G_bands', 'Network Support']].head(10))
```

### 7. Extracting Chip Brand, CPU Cores, and CPU Clock Speed

```python
def extract_chip_info(row):
    cpu_str = str(row['CPU']).strip() if pd.notna(row['CPU']) else ""
    chip_str = str(row['Chipset']).strip() if pd.notna(row['Chipset']) else ""
    combined_upper = (chip_str + " " + cpu_str).upper()

    # 1. Extract Chipset Brand
    if "SNAPDRAGON" in combined_upper or "QUALCOMM" in combined_upper:
        brand = "Snapdragon"
    elif any(k in combined_upper for k in ["MEDIATEK", "HELIO", "MT6", "MT8"]):
        brand = "MediaTek"
    elif "EXYNOS" in combined_upper:
        brand = "Exynos"
    elif "APPLE" in combined_upper or str(row['brand']).strip().upper() == "APPLE":
        brand = "Apple"
    elif "KIRIN" in combined_upper or "HISILICON" in combined_upper:
        brand = "Kirin"
    elif "SPREADTRUM" in combined_upper:
        brand = "Spreadtrum"
    elif "INTEL" in combined_upper or "ATOM" in combined_upper:
        brand = "Intel"
    elif pd.isna(row['Chipset']):
        brand = None
    else:
        brand = "Other"

    # 2. Extract CPU Core Count (Cores)
    cpu_upper = cpu_str.upper()
    if not cpu_str:
        cores = None
    elif "DECA-CORE" in cpu_upper or "10-CORE" in cpu_upper:
        cores = "10 cores"
    elif "OCTA-CORE" in cpu_upper or "8-CORE" in cpu_upper:
        cores = "8 cores"
    elif "HEXA-CORE" in cpu_upper or "6-CORE" in cpu_upper:
        cores = "6 cores"
    elif "QUAD-CORE" in cpu_upper or "4-CORE" in cpu_upper:
        cores = "4 cores"
    elif "DUAL-CORE" in cpu_upper or "2-CORE" in cpu_upper:
        cores = "2 cores"
    else:
        cores = "1 core"

    # 3. Extract maximum CPU clock speed (GHz)
    ghz_matches = re.findall(r'(\d+(?:\.\d+)?)\s*GHz', cpu_str, re.IGNORECASE)
    if ghz_matches:
        clock = f"{max([float(x) for x in ghz_matches])} GHz"
    else:
        mhz_matches = re.findall(r'(\d+(?:\.\d+)?)\s*MHz', cpu_str, re.IGNORECASE)
        clock = f"{round(max([float(x) for x in mhz_matches]) / 1000.0, 2)} GHz" if mhz_matches else None

    return brand, cores, clock

# Apply the function to create 3 processor specification columns
chip_extracted = df.apply(extract_chip_info, axis=1)
df['Chip Brand'] = [x[0] for x in chip_extracted]
df['CPU Cores'] = [x[1] for x in chip_extracted]
df['CPU Clock Speed'] = [x[2] for x in chip_extracted]

display(df[['Chipset', 'CPU', 'Chip Brand', 'CPU Cores', 'CPU Clock Speed']].head(10))
```

### Extracting Release Year, Standardizing NFC, and Dropping Unnecessary Raw Text Columns
We extract the 4-digit release year from the `announced` column, binary-encode the `NFC` column (`1`: supported, `0`: not supported), standardize the brand and price columns, drop the initial raw text columns that are no longer needed for analysis, and reset the DataFrame index.

```python
import pandas as pd

# Reset index to default 0-based indexing
df_final = df.reset_index(drop=True)

# 1. Process the announced column (extract 4-digit year only)
# If no year is found, it will automatically become NaN
df_final['Release Year'] = df_final['announced'].astype(str).str.extract(r'(\d{4})').astype(float)

# 2. Process the NFC column (Has data -> 1, NaN -> 0)
df_final['NFC_Clean'] = df_final['NFC'].notna().astype(int)

# Standardize Brand, Price (EUR), and reassign the cleaned RAM column
df_final['Brand'] = df_final['brand'].astype(str).str.strip()
df_final['Price (EUR)'] = df_final['approx_price_EUR']
df_final['RAM'] = df_final['RAM_Clean']

# Define the list of unnecessary raw text columns to drop
# Included original 'announced' and 'NFC' since we created cleaned versions above
cols_to_drop = [
    'brand', 'model', 'network_technology', '2G_bands', '3G_bands', '4G_bands',
    'network_speed', 'GPRS', 'EDGE', 'status', 'dimentions',
    'weight_g', 'weight_oz', 'SIM', 'display_type', 'display_resolution',
    'display_size', 'OS', 'CPU', 'Chipset', 'GPU', 'memory_card',
    'internal_memory', 'RAM_Clean', 'primary_camera', 'secondary_camera',
    'loud_speaker', 'audio_jack', 'WLAN', 'bluetooth', 'GPS',
    'radio', 'USB', 'sensors', 'battery', 'colors', 'approx_price_EUR', 'img_url',
    'announced', 'NFC'
]

df_final = df_final.drop(columns=cols_to_drop, errors='ignore')

print(f"Dataset shape after dropping raw columns: {df_final.shape}")
pd.set_option('display.max_columns', None)
display(df_final.head(10))
```

**Next, we will gather and reorder all extracted features into a new clean DataFrame**

```python
# Reorder and gather the standardized and extracted feature columns
selected_features = [
    'Brand', 'Price (EUR)',
    'RAM', 'ROM',
    'Screen Size', 'Screen-to-Body Ratio', 'Display Panel', 'Display Resolution',
    'Rear Camera', 'Front Camera',
    'Battery Capacity', 'Battery Design',
    'Thickness', 'Weight',
    'Fingerprint Sensor', 'Network Support',
    'Chip Brand', 'CPU Cores', 'CPU Clock Speed',
    'Release Year', 'NFC_Clean'
]

# Create a new DataFrame containing the processed features
df_features = df_final[selected_features].copy()

print(f"New feature table shape: {df_features.shape}")
display(df_features.head(5))
```

**For columns containing measurement units, we will strip the unit strings and convert them to numeric values**

```python
import pandas as pd
import numpy as np

# Create a copy to avoid SettingWithCopyWarning
df_features = df_features.copy()

# Use Regex to strip all unit characters and whitespace, keeping only digits (0-9) and decimal points (.) for safe numeric conversion
unit_columns = [
    'RAM', 'ROM', 'Screen Size', 'Screen-to-Body Ratio', 
    'Rear Camera', 'Front Camera', 'Battery Capacity', 
    'Thickness', 'Weight', 'CPU Cores', 'CPU Clock Speed'
]

for col in unit_columns:
    if col in df_features.columns:
        # Remove unit characters and whitespace using Regex
        df_features[col] = df_features[col].astype(str).str.replace(r'[^\d.]+', '', regex=True)
        # Convert to numeric; 'nan' or invalid strings become NaN
        df_features[col] = pd.to_numeric(df_features[col], errors='coerce')

# Rename columns to include their original measurement units in parentheses
rename_dict = {
    'RAM': 'RAM (GB)',
    'ROM': 'ROM (GB)',
    'Screen Size': 'Screen Size (inch)',
    'Screen-to-Body Ratio': 'Screen-to-Body Ratio (%)',
    'Rear Camera': 'Rear Camera (MP)',
    'Front Camera': 'Front Camera (MP)',
    'Battery Capacity': 'Battery Capacity (mAh)',
    'Thickness': 'Thickness (mm)',
    'Weight': 'Weight (g)',
    'CPU Cores': 'CPU Cores',
    'CPU Clock Speed': 'CPU Clock Speed (GHz)'
}
df_features = df_features.rename(columns=rename_dict)

# Display the result after stripping units
display(df_features.head(15))

# Print DataFrame info
df_features.info()
```

**For the display resolution column, we will parse the pixel dimensions and calculate the diagonal pixel count (Pythagorean theorem), making it convenient to compute PPI later**

```python
import re
import pandas as pd
import numpy as np

def calculate_diagonal_pixels(resolution_str):
    if pd.isna(resolution_str):
        return np.nan
    # Find numeric patterns of the form: width x height or width X height
    match = re.search(r'(\d+)\s*[xX]\s*(\d+)', str(resolution_str))
    if match:
        width = int(match.group(1))
        height = int(match.group(2))
        # Pythagorean theorem: sqrt(width^2 + height^2)
        return np.sqrt(width**2 + height**2)
    return np.nan

# Apply the function to calculate diagonal resolution in pixels and drop the original string column
if 'Display Resolution' in df_features.columns:
    df_features['Diagonal Resolution (Pixels)'] = df_features['Display Resolution'].apply(calculate_diagonal_pixels)
    df_features = df_features.drop(columns=['Display Resolution'])

display(df_features[['Brand', 'Diagonal Resolution (Pixels)']].head(10))
print(df_features['Diagonal Resolution (Pixels)'].dtype)
```
